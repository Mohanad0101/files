# 8. Output and f-strings

## Understand

`print()` displays values. With multiple arguments, a space separates them by default.

```python
print("A", "B", sep=" -> ")
print("same line", end=" ")
print("continued")
```

For readable dynamic text, modern Python commonly uses **f-strings**:

```python
name = "Ada"
score = 92.5
```

</div>
